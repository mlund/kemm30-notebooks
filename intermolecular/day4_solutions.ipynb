## The Born Energy

Transfer of an ion from one medium to another is associated with a free energy change due to a change in the polarization of the surrounding medium. This is the Born energy:
$$
u_{born} = \frac{q^2}{8 \pi \epsilon_0 a} \cdot \left ( \frac{1}{\epsilon_2} - \frac{1}{\epsilon_1}  \right )
$$

We will here derive that, starting from a molecular perspective and assume that the ion interact with cloud of polarizable molecules. The interaction is given by an ion-induced dipole interactions, which decays as $1/r^4$.

```python
q, e, rho, pi, r, alpha, a, epsilon_0, epsilon_r = symbols("q e rho pi r alpha a epsilon_0 epsilon_r", real=True, positive=True)
u_pol = -q**2 * alpha / 2 / (4 * pi * epsilon_0 * epsilon_r * r**2)**2
Eq(symbols("u_{pol}"), u_pol)
```

If the radius of the ion is $a$ and the number density of polarizable molecules outside the ion is $\rho$, we can sum up all the interactions using a volume integral. We also assume that $g(1)=1$ for $r>a$ which we also used when calculating the mean energy for a Lennard-Jones fluid.

```python
# See Israelachvili p98
u = rho * 4 * pi  * integrate(r**2 * u_pol, (r, a, oo))
Eq(symbols("u"), u)
```

The [electric susceptibility](https://en.wikipedia.org/wiki/Electric_susceptibility), $\chi$ is related to $\rho \alpha / \epsilon_0$ which in turn is related to the relative permittivity, $\chi = \epsilon_r -1$. It follows that $d\chi = d\epsilon_r$.

To show that $d\chi = d\epsilon_r$, differentiate  $\chi = \epsilon_r -1$ with respect to any independent variable, say $t$:
$$
\frac{d\chi}{dt} = \frac{d\epsilon_r}{dt} - \frac{d1}{dt}
\Rightarrow d\chi = d\epsilon_r
$$
</details>
To get the energy from moving from one medium to another, we can now integrate to get the so-called Born energy:
$$
\Delta u = -\int_{\chi_1}^{\chi_2} \frac{q^2}{8\pi\epsilon_0\epsilon_r^2 a} d\chi
= -\int_{\epsilon_{r,1}}^{\epsilon_{r,2}} \frac{q^2}{8\pi\epsilon_0\epsilon_r^2 a} d\epsilon_r
= \frac{q^2}{8\pi\epsilon_0a}\left ( \frac{1}{\epsilon_{r,1}} - \frac{1}{\epsilon_{r,2}} \right )
$$ 

More generally, each of the two $\frac{q^2}{8\pi\epsilon_0\epsilon_r a}$ terms can be regarded as a self-energy for charging up a sphere in a specific dielectric medium. With this in mind we now get to the interesting conclusion that the ion-ion interaction in water can be regarded as a _change in solvation_ as a the two ions approach.

The interaction in water can be found by considering the process of
1. moving two ions, $q_1=q_2$ from infinity to contact so that $r=a$,
2. transferring both ions to the aqueous phase.

The total free energy is then

$$
u_w(r) = \frac{q^2}{4\pi\epsilon_0 a} + 2 \cdot \frac{q^2}{8\pi\epsilon_0 a} \left ( \frac{1}{\epsilon_r} - 1\right )
= \frac{q^2}{4\pi\epsilon_0\epsilon_r a}
$$
where we see that the well-known Coulomb energy in a medium is recovered.
